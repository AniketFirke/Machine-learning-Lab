```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_performance_1200.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Number of records:", df.shape[0])
print("Number of attributes:", df.shape[1])
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

print("Missing values after preprocessing:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
mean_values = df.mean(numeric_only=True)

print("Mean:")
print(mean_values)
```

```python
median_values = df.median(numeric_only=True)

print("Median:")
print(median_values)
```

```python
mode_values = df.mode(numeric_only=True).iloc[0]

print("Mode:")
print(mode_values)
```

```python
min_values = df.min(numeric_only=True)

print("Minimum:")
print(min_values)
```

```python
max_values = df.max(numeric_only=True)

print("Maximum:")
print(max_values)
```

```python
q1 = df.quantile(0.25, numeric_only=True)

print("First Quartile (Q1):")
print(q1)
```

```python
q3 = df.quantile(0.75, numeric_only=True)

print("Third Quartile (Q3):")
print(q3)
```

```python
iqr = q3 - q1

print("Interquartile Range (IQR):")
print(iqr)
```

```python
variance = df.var(numeric_only=True)

print("Variance:")
print(variance)
```

```python
std_dev = df.std(numeric_only=True)

print("Standard Deviation:")
print(std_dev)
```

```python
covariance = df[["Study_Hours", "Final_Score"]].cov()

print("Covariance Matrix:")
print(covariance)
```

```python
correlation = df["Study_Hours"].corr(
    df["Final_Score"],
    method="pearson"
)

print("Pearson Correlation Coefficient:")
print(correlation)
```

```python
correlation_matrix = df.corr(numeric_only=True)

print(correlation_matrix)
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours"],
    df["Final_Score"],
    alpha=0.6
)

plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Study Hours vs Final Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Study_Hours"],
    bins=20,
    edgecolor="black"
)

plt.xlabel("Study Hours")
plt.ylabel("Frequency")
plt.title("Distribution of Study Hours")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Final_Score"],
    bins=20,
    edgecolor="black"
)

plt.xlabel("Final Score")
plt.ylabel("Frequency")
plt.title("Distribution of Final Scores")

plt.show()
```

```python
columns = [
    "Age",
    "Study_Hours",
    "Attendance_Percent",
    "Assignments_Score",
    "Midterm_Score",
    "Final_Score",
    "Projects_Score"
]

for column in columns:
    plt.figure(figsize=(7, 4))
    plt.hist(df[column], bins=20, edgecolor="black")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title("Histogram of " + column)
    plt.show()
```

```python
plt.figure(figsize=(10, 6))

df[
    [
        "Study_Hours",
        "Attendance_Percent",
        "Assignments_Score",
        "Midterm_Score",
        "Final_Score",
        "Projects_Score"
    ]
].boxplot()

plt.title("Box Plot of Student Performance Variables")
plt.ylabel("Value")

plt.xticks(rotation=30)

plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.boxplot(df["Final_Score"])

plt.ylabel("Final Score")
plt.title("Box Plot of Final Score")

plt.show()
```

```python
Q1 = df["Final_Score"].quantile(0.25)
Q3 = df["Final_Score"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Final_Score"] < lower_limit) |
    (df["Final_Score"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of potential outliers:", len(outliers))
```
